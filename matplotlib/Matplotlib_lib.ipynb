# **Matplotlib**

Matplotlib is a Python library used to create graphs and visualizations like line charts, bar charts, pie charts, etc.


__workflow__

1.Prepare Data

2.Plot Chart:

	a.Line Chart

	b.Bar Chart

	c.Pie Chart

	d.Scatter Chart

	e.Histogram

3.Customize the looks of chart and give labels(colors,font,legends,title)

4.Show Graph

```python
import matplotlib.pyplot as plt #pypot is a module of matplotlib used for plotting
```

### Linechart
(show trend)

```python
x=[0,1,2,3,4,5,6]        #x = horizontal values
y1=[0,10,20,30,40,50,60]   #y1 = vertical values
plt.plot(x,y1)        #plot() = creates graph
plt.show()           #show() = displays graph
```

```python
plt.figure(figsize=(5,5))#background 
plt.plot(x,y1)
plt.title("My First Graph",fontsize=18,color="green",fontweight="bold")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.plot(x,y1,linestyle="--",marker="o",linewidth=3)#linestyle='--' → dashed line
plt.title("Styled Graph")#marker='o' → points shown
plt.xticks(fontsize=10,rotation=30)
plt.yticks(fontsize=10,rotation=45)
plt.show()
```

```python
y2=y1[::-1]
plt.plot(x,y1,label="line 1")
plt.plot(x,y2,label="line 2")

intersecting_point=((x[0]+x[-1])/2,(y1[0]+y1[-1])/2)#only for this graph,(3,30)
plt.text((x[0]+x[-1])/2,(y1[0]+y1[-1])/2,f"Common point:{intersecting_point}",ha="left",va='center',fontsize=12,color="red")

plt.legend(loc="upper right")#It creates a small box on the graph that tells:which line represents what
plt.show()
```

```python
import numpy as np
plt.axhline(np.mean(x),linestyle="--",label=f"Mean={np.mean(x):.2f}")#axhline:Draw a horizontal line across the graph
plt.legend(loc="upper right",frameon=False)#frameon=appeareance of legend box
plt.grid()#make these boxes
ax=plt.gca()#get current axis
ax.set_facecolor("lightgray")
ax.set_xlim(0, 60)#axis control
ax.set_ylim(0, 6)
```

### Bar chart
(categories vs values comparison)

```python
subjects=["Maths","English","Hindi"]
marks=[80,70,90]
plt.bar(subjects,marks)#categories on X-axis,Values on Y axis
plt.show()
```

```python
plt.bar(subjects,marks,color="red")
plt.title("Students Marks",fontsize=30)
plt.xlabel("Subjects",fontsize=18)
plt.ylabel("Marks",fontsize=18)
plt.show()
```

```python
plt.bar(subjects,marks,width=0.5,color=["#4CAF50","red","blue"],edgecolor="black")
plt.grid(axis='y',linestyle="--",alpha=0.5)#alpha=transparency level of grid
plt.show()
```

```python
plt.barh(subjects,marks)#Horizontal Bar Chart use when categories have long name
plt.show()
```

```python
import numpy as np
x=np.arange(len(subjects))
x
```

```python
# Multiple bar charts (Comparison)
import numpy as np
x=np.arange(len(subjects))#[0, 1, 2]
marks1=[80,70,90]
marks2=[60,75,85]
plt.bar(x-0.2,marks1,width=0.4,label="Student 1")
plt.bar(x+0.2,marks2,width=0.4,label="Student 2")
plt.xticks(x,subjects,rotation=30)#to prevent overlap use xticks
plt.legend()
plt.grid()
plt.show()
```

### Pie Charts
(parts of total 100% or show composition)



```python
subjects=["Maths",'English',"Hindi"]
marks=[30,50,20]
plt.pie(marks)
plt.show()
```

```python
exp=[0.1,0.2,0]
plt.pie(marks,labels=subjects,explode=exp,shadow=True)
plt.title("Marks Distribution")
plt.show()
```

```python
plt.pie(marks,labels=subjects,autopct="%1.1f%%",colors=["green","yellow","pink"],startangle=0)#Double %% shows real percentage sign here
plt.show()
```

## Scatter Plot 
dots on graphs without line(Show relationships)


```python
x=[1,2,3,4,5]
y=[i*10 for i in x ]
plt.scatter(x,y,color="red",s=100,alpha=0.6,marker="p")#s=size of points,alpha=transperancy
plt.title("Study vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
for i in range(len(x)):
    plt.text(x[i], y[i], str(y[i]))
plt.show()
```

### Histogram
(range me kitni values hain)

```python
marks=[10,10,40,27,40,40,90,60]#Height of bars = count
plt.hist(marks,bins=10,color="blue",edgecolor="black",rwidth=0.2)#bin=no. of ranges ((90-10)/10)=8(band width)(10–18 18–26 26–34 34–42 42–50 50–58 58–66 66–74 74–82 82–90)
plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.show()
```

### Subplot

```python
# Subplot:A subplot means multiple graphs inside ONE figure
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("Line")

plt.subplot(1,2,2)
plt.bar(x,y1)
plt.title("Bar")
plt.show()

```

```python
# Subplots=By using subplots we can control each part
fig, ax=plt.subplots(2,2)#Here (2,2)  is no. of rows and no.of columns
ax[0][0].plot(x,y1)
ax[0][1].bar(x,y1)
ax[1][0].plot(x,y2)
ax[1][1].bar(x,y2)
```
