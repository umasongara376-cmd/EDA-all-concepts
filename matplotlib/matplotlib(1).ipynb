```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
points = np.array([3,8,1,10])
plt.plot(points, marker = "*", alpha = 0.9 , color = "red" , linestyle="solid")   # alpha use for transparancy of graph it is btw 0-1

plt.show()
```

```python
i=np.array([3,8,1,10,7])
plt.plot(i, "k--h", ms=15, mec="red" , mfc="pink")   # ms is the marker size ,  mec is the marker edge color, mfc means marker face color
plt.show()
```

```python
i=np.array([3,8,1,10])
j=np.array([6,2,8,1])

ax=plt.gca()   #get current axes
ax.set_facecolor("pink")

plt.plot(i, color="purple" , marker="*" , ms=15 , mec="black" , mfc="white")
plt.plot(j, color="green" , marker="h" , ms=10 , mec="black" , mfc="white")

plt.show()
```

```python
a1=np.array([3,8,1,10])
a2=np.array([6,2,8,1])

ax=plt.gca()   #get current axes
ax.set_facecolor("black")


plt.plot(a1,label="uma", color="orange" , marker="*" , ms=15 , mec="red" , mfc="white")
plt.plot(a2, label="vishwas",color="yellow" , marker="h" , ms=10 , mec="red" , mfc="white")

plt.xlabel("Average Pulse" , color="red" , fontsize=10)
plt.ylabel("Calorie Burnage" , color="blue" , fontsize=10)
plt.title("Line Plot" , color="green" , fontsize=30)

plt.legend(loc="lower left")
plt.grid(axis="y")
plt.show()
```

```python
months =["Jan","Feb","March","April","May","June"]
sales=[120,150,180,170,210,250]

plt.figure(figsize=(8,5))

plt.plot(months,sales,marker="o",linestyle="--",linewidth=2,markersize=8,label="sales",mfc="pink",mec="red",color="purple")

plt.title("Monthly Sales Trend",fontsize=15,fontweight="bold",color="blue",loc="center")
plt.xlabel("Months",fontsize=10,fontweight="normal",color="red",loc="center")
plt.ylabel("Sales",fontsize=10,fontweight="normal",color="red",loc="center")
plt.grid(axis="both",linestyle="--",linewidth="0.4",color="green",alpha=0.5)
plt.legend(loc="best",fontsize=14,title="Sales",shadow=True,facecolor="#B57474",frameon=True,framealpha=0.9,edgecolor="#B3BD84")


plt.show()
```

### scatter plot

when we want to find relationship we use this graph or for scattered data

```python
x=np.array([1,2,3,4,5,6])
y=np.array([1,8,27,64,125,216])
colors=np.array([0,10,20,30,40,45])

plt.scatter(x,y,c=colors,cmap="plasma")  #YLOrBr,viridis,cividis,plasma,megma,coolwarm,

plt.colorbar()
plt.show()
```

```python
x=np.array([1,2,3,4,5,6])
y=np.array([1,8,27,64,125,216])
colors=np.array([0,10,20,30,40,45])

plt.scatter(x,y,c=colors,cmap="YlOrBr")  #YLOrBr,viridis,cividis,plasma,megma,coolwarm,

plt.colorbar()
plt.show()
```

```python
experience=[1,2,3,4,5,6,7,8]
salary=[25,30,35,40,45,50,60,70]

plt.Figure(figsize=(10,8))

plt.scatter(experience,salary , s=100,alpha=0.8 , marker="v", color="#295A3E")

plt.title("Experience vs Salary")
plt.xlabel("EXperience")
plt.ylabel("Salary")

plt.grid(axis="x",alpha=0.3)
plt.show()
```

### Box plot

```python
salary=[25,28,30,32,35,200]

plt.figure(figsize=(18,5))  # 18 is width of figure, 5 is height of figure
plt.boxplot(salary, patch_artist=True , showmeans=True , vert=False )  

plt.title("Salary Distributioin" , color="#660C44" , fontsize=20 , fontfamily="serif", loc="center")

plt.show()
```

```python
years=[2021,2022,2023,2024,2025]
revenue=[100,130,100,210,500]

plt.figure(figsize=(8,5))

plt.fill_between(
    years,
    revenue,
    alpha=1,
    color="#409969",
    edgecolor="#BB2F2F",
    linewidth=2
)

plt.title("Revenue Growth" , color="#652C2C")
plt.xlabel("years" , color="#672546" )
plt.ylabel("Revenue" , color="#41185B")

plt.show()
```

```python
d=[40,30,20,10]
labels=["Python","Java","C++","JavaScript"]

plt.figure(figsize=(5,6))

plt.pie(d, labels=labels, shadow=True, explode=[0.03,0.05,0.09,0.05] , autopct="%0.1f%%" , colors=["pink" , "purple" , "blue" , "green" ])
```

```python
d=[40,30,20,10]
labels=["Python","Java","C++","JavaScript"]

plt.figure(figsize=(8,8))

plt.pie(d, labels=labels, shadow=True, explode=[0.03,0.05,0.07,0.05] , autopct="%0.1f%%" , colors=["hotpink" , "green" , "purple" , "#37A094" ],
        startangle=90, counterclock=True, wedgeprops={"edgecolor":"#0E4D42", "linewidth":2.0},
        textprops={"fontsize":10,"fontweight":"bold"})

plt.title("Programming Language Popularity")

plt.legend(title="Language",loc="lower left")

plt.show()
```
