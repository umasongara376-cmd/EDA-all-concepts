### loc(local based)

```python
df=pd.read_csv(r'C:\Users\umara\Downloads\student.csv')
df
```

```python
df.loc[1,'Age'] #it will give the value of age column in 1st row
```

```python
df.loc[3,'name'] #it will give the value of name column in 3rd row
```

```python
df.loc[4] #it will give the value of all columns in 4th row
```

```python
df.loc[1:3,['Age','city']] #it will give the value of age and city column in 1st to 3rd row
```

```python
df.loc[4,'name']
```

```python
df.loc[0:4,["marks"]]
```

### iloc(index based) 
##### it uses integer position

```python
df.iloc[0:4,2] #it will give the value of marks column in 0 to 4 row
```

```python
df.iloc[1,4] #it will give the value of city column in 1st row
```

```python
df.iloc[0:3,0:4] #it will give the value of all columns in 0 to 3 row,here 0:3 means 0 to 2 row and 0:4 means 0 to 3 column
```

##### difference b/w loc and iloc

```python
df.loc[0:3] #it include 0 to 3 row and all columns
```

```python
df.iloc[0:3] # it include 0 to 2 row and all columns
```

### filtering
##### this is where analysis begins

```python
df[df["Age"]>30] #it will give the value of all columns where age is greater than 30
```

```python
df[df["grade"]=='A'] #it will give the value of all columns where grade is A
```

```python
df[df["marks"]<90] #it will give the value of all columns where marks is less than 90
```

```python
df[(df["Age"]>30) & (df["grade"]=='A')] #it will give the value of all columns where age is greater than 30 and grade is A  
```

```python
df[(df["marks"]>80) & (df["grade"]=='A')] #it will give the value of all columns where marks is greater than 80 and grade is A    
```

```python
df[(df["marks"]>80) | (df["grade"]=='A')] # it will give the value of all columns where marks is greater than 80 or grade is A
```

### data cleaning

```python
data={'name':['uma','vedik','vijay','komal','ajay','riya'],
        'marks':[64,98,None,67,89,90],
        'grade':[None,'A+','B','B+',None,'A+']}   
df=pd.DataFrame(data)
df
```

```python
df.isnull() # it will give the boolean value true for null value and false for non null value
```

```python
df.isnull().sum() # it will give the count of null value in each column
```

#### drop  missing value

```python
df.dropna() # it will drop the row which contain null value
```

```python
df.dropna(subset=["marks"]) # it will drop the row which contain null value in marks column but it will not drop the row which contain null value in grade column
```

### fill missing value

```python
import numpy as np
df["marks"]=df["marks"].fillna(np.mean(df["marks"])) # it will fill the null value in marks column with mean value of marks column
df
```

```python
df['city']=['jaipur','delhi','indore','pune','ujjain','bhopal']
df
```

```python
df.fillna({'marks': 0, 'grade': 'Unknown'}, inplace=True) # it will fill the null value with 0 in marks column and unknown in grade column
df
```

```python
df['name']=df['name'].fillna('Unknown') # it will fill the null value in name column with unknown
df
```

```python
df.duplicated() # it will give the boolean value true for duplicate value and false for non duplicate value
```
