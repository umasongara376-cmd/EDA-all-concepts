```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data= np.random.normal(170,10,1000)

sns.histplot(data=data,bins=20,kde=True,color="hotpink")   # kde is used for line show on columns

plt.title("Height Distribution")
plt.show()
```

```python
df = pd.DataFrame({
    "Experience":[1,2,3,4,5,6,7],
    "Salary":[25,30,40,50,60,70,85]
})

sns.scatterplot(data=df, x="Experience" , y="Salary" , s=200 , hue="Salary"  ,style="Salary" , palette="Set1" ) # hue is used for color , style is used for shape of dot , palette is used for color theme

plt.title("Experience vs Salary")
plt.show()
```

```python
df = pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Salary":[100,120,150,130,180]
    
})

sns.lineplot(
    data=df, x="Month", y="Salary", marker="o", mec="k", mfc="#4BA777", ms=10, linewidth=2, linestyle="--", color="#782424"
    )

plt.title("Sales Trend")
plt.show()
```

```python
df= pd.DataFrame({
    "Course":["Python","Java","C++","SQL"],
    "Students":[80,60,40,50]
})

sns.barplot(
    data=df, x="Course", y="Students", edgecolor="k", hue="Course", palette="Set2"
)

plt.title("Students by Course")
plt.show()
```

```python
salary=[-100,25,28,30,35,40,50,55,60,100,200]

sns.boxplot(
    data=salary, showmeans=True
)

plt.title("Salary Distribution")
plt.show()
```

```python
salary=[-100,25,28,30,35,40,50,55,60,100,200]

sns.boxenplot(
    data=salary
)

plt.title("Salary Distribution")
plt.show()
```

```python
df=pd.DataFrame({
    "Gender":["Male","Female","Male","Female","Male"]
})

sns.countplot(
    data=df,x="Gender" , color="#1E8387" , hue="Gender" , palette="Set2"
)

plt.title("Gender Count", fontsize=20 )
plt.show()
```

```python
df=pd.DataFrame({
    "Experience":[20,25,30,35,40],
    "Salary":[25,35,45,55,65],
    "Age":[21,33,45,57,29],
    "Marks":[89,99,26,27,88]
})

sns.heatmap(
    df.corr(),annot=True,cmap="brg",linewidths=0.1
)

plt.title("Correlation Heatmap")
plt.show()

# cmap is an color palate
```
